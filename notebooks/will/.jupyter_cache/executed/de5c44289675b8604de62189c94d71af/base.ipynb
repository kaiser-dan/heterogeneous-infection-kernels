```python
#| label: load-posteriors
def load_mu_samples(experiment, model):
    """Load non-parametric kernel mode posterior samples (mu[1] for K=1)."""
    path = RESULTS / experiment / model / "posterior_samples.parquet"
    draws = pd.read_parquet(path)
    mu_cols = sorted([c for c in draws.columns if c.startswith("mu[1,")],
                     key=lambda c: int(c.split(",")[1].rstrip("]")))
    mu_samples = draws[mu_cols].values
    return mu_samples, draws

def load_rho_samples(experiment):
    """Load simple contagion rho posterior samples."""
    path = RESULTS / experiment / "simple_contagion" / "posterior_samples.parquet"
    draws = pd.read_parquet(path)
    rho = draws["rho[1]"].values
    return rho

# Non-parametric monotonic posteriors
mu_np_exp1, _ = load_mu_samples("experiment_1_homogeneous", "non_parametric_monotonic")
mu_np_exp2, _ = load_mu_samples("experiment_2_sigmoid", "non_parametric_monotonic")
mu_np_exp3, _ = load_mu_samples("experiment_3_mixture", "non_parametric_monotonic")

# Simple contagion rho posteriors
rho_exp1 = load_rho_samples("experiment_1_homogeneous")
rho_exp2 = load_rho_samples("experiment_2_sigmoid")
rho_exp3 = load_rho_samples("experiment_3_mixture")

kmax = meta["kmax"]
nu = np.arange(1, kmax + 1)

print(f"Loaded {mu_np_exp1.shape[0]} samples per experiment (kmax={kmax})")
print(f"  Exp 1: ρ̂ = {rho_exp1.mean():.4f} (truth: {meta['BETA']})")
print(f"  Exp 2: ρ̂ = {rho_exp2.mean():.4f} (best simple fit to sigmoid)")
print(f"  Exp 3: ρ̂ = {rho_exp3.mean():.4f} (best simple fit to mixture)")
```

```python
#| label: mahalanobis
def compute_mahalanobis_decomposed(mu_samples, rho_samples):
    """
    Compute Mahalanobis distance and decompose into spread + offset.

    Returns:
        D_total: per-sample D_M (distance from SC reference)
        D_centered: per-sample D_M centered on posterior mean (~chi2_k)
        lam: non-centrality parameter (offset of posterior mean from SC ref)
        mu_sc: simple contagion reference kernel
    """
    n_samples, k = mu_samples.shape
    nu = np.arange(1, k + 1)

    mu_bar = mu_samples.mean(axis=0)
    Sigma = np.cov(mu_samples, rowvar=False)
    Sigma += 1e-8 * np.eye(k)
    L = np.linalg.cholesky(Sigma)

    rho_hat = np.mean(rho_samples)
    mu_sc = 1.0 - (1.0 - rho_hat) ** nu

    # Offset (non-centrality parameter)
    z_offset = np.linalg.solve(L, mu_bar - mu_sc)
    lam = z_offset @ z_offset

    D_total = np.zeros(n_samples)
    D_centered = np.zeros(n_samples)
    for s in range(n_samples):
        z_total = np.linalg.solve(L, mu_samples[s] - mu_sc)
        D_total[s] = z_total @ z_total
        z_cent = np.linalg.solve(L, mu_samples[s] - mu_bar)
        D_centered[s] = z_cent @ z_cent

    return D_total, D_centered, lam, mu_sc

D_t1, D_c1, lam1, mu_sc1 = compute_mahalanobis_decomposed(mu_np_exp1, rho_exp1)
D_t2, D_c2, lam2, mu_sc2 = compute_mahalanobis_decomposed(mu_np_exp2, rho_exp2)
D_t3, D_c3, lam3, mu_sc3 = compute_mahalanobis_decomposed(mu_np_exp3, rho_exp3)

print(f"{'Experiment':<20} {'λ (offset)':>10} {'Median D_total':>15} {'Median D_centered':>18}")
print("-" * 65)
for label, lam, Dt, Dc in [("Exp 1 (simple)", lam1, D_t1, D_c1),
                             ("Exp 2 (sigmoid)", lam2, D_t2, D_c2),
                             ("Exp 3 (mixture)", lam3, D_t3, D_c3)]:
    print(f"{label:<20} {lam:>10.1f} {np.median(Dt):>15.1f} {np.median(Dc):>18.1f}")
print(f"\nχ²({kmax}) median = {stats.chi2.median(kmax):.1f}")
```

```python
#| label: calibrated-test
if has_ppc:
    test_results = {}
    observed_lambdas = {"Exp 1 (simple)": lam1, "Exp 2 (sigmoid)": lam2, "Exp 3 (mixture)": lam3}

    print(f"{'Experiment':<20} {'λ_obs':>8} {'λ_null median':>14} {'λ_null 95th':>12} {'p_ppc':>8} {'Decision':>20}")
    print("-" * 85)
    for name in ["Exp 1 (simple)", "Exp 2 (sigmoid)", "Exp 3 (mixture)"]:
        lam_obs = observed_lambdas[name]
        if name in ppc_lambdas:
            lam_null_dist = ppc_lambdas[name]
            p_ppc = np.mean(lam_null_dist >= lam_obs)
            decision = "FAIL TO REJECT H₀" if p_ppc > 0.05 else "REJECT H₀"
            test_results[name] = (p_ppc, decision)
            print(f"{name:<20} {lam_obs:>8.1f} {np.median(lam_null_dist):>14.1f} "
                  f"{np.percentile(lam_null_dist, 95):>12.1f} {p_ppc:>8.3f} {decision:>20}")
        else:
            test_results[name] = (np.nan, "NO PPC DATA")
            print(f"{name:<20} {lam_obs:>8.1f} {'—':>14} {'—':>12} {'—':>8} {'NO PPC DATA':>20}")
else:
    print("PPC data not available. Run scripts/simplicity_test_sim.py first.")
```

```python
#| label: explained-variance
def compute_simplicity_index(mu_samples, rho_samples):
    """
    Compute cosine-similarity simplicity index S = cos²θ in whitened space.
    """
    k = mu_samples.shape[1]
    nu = np.arange(1, k + 1)

    mu_bar = mu_samples.mean(axis=0)
    Sigma = np.cov(mu_samples, rowvar=False)
    Sigma += 1e-8 * np.eye(k)
    Sigma_inv = np.linalg.inv(Sigma)

    rho_hat = np.mean(rho_samples)
    mu_sc = 1.0 - (1.0 - rho_hat) ** nu

    cross = mu_bar @ Sigma_inv @ mu_sc
    L_np = mu_bar @ Sigma_inv @ mu_bar
    L_sc = mu_sc @ Sigma_inv @ mu_sc
    cos_theta = cross / np.sqrt(L_np * L_sc)
    S = cos_theta ** 2

    return S, cos_theta, L_np, L_sc

S1, cos1, Lnp1, Lsc1 = compute_simplicity_index(mu_np_exp1, rho_exp1)
S2, cos2, Lnp2, Lsc2 = compute_simplicity_index(mu_np_exp2, rho_exp2)
S3, cos3, Lnp3, Lsc3 = compute_simplicity_index(mu_np_exp3, rho_exp3)

print(f"{'Experiment':<20} {'cosθ':>8} {'S = cos²θ':>12} {'L_np':>10} {'L_sc':>10}")
print("-" * 62)
for label, cos, S, Lnp, Lsc in [("Exp 1 (simple)", cos1, S1, Lnp1, Lsc1),
                                  ("Exp 2 (sigmoid)", cos2, S2, Lnp2, Lsc2),
                                  ("Exp 3 (mixture)", cos3, S3, Lnp3, Lsc3)]:
    print(f"{label:<20} {cos:>8.4f} {S:>12.4f} {Lnp:>10.1f} {Lsc:>10.1f}")
```
